## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research Question

Which content pages should be prioritized for refresh or review using observable search and engagement signals?

### Decision

The project helps the SEO or content team decide which pages should be reviewed first.

### Action

Based on the recommendations, the team can refresh, expand, monitor, protect, or further investigate the selected pages.

### Cost of a Wrong Recommendation

If an important page is ranked too low, valuable opportunities to improve traffic may be missed. If a low-impact page is ranked too high, the team may spend time and effort on pages that provide little benefit.

### Why Machine Learning?

Page performance depends on several signals rather than a single metric. Machine learning can combine these observable signals to produce a ranked recommendation that supports human decision-making while remaining interpretable.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains **30,000 pages** and **44 features**, providing enough information to explore search and content performance.

Three observations support my choice of this lane:

- Search volume has a very weak correlation with impressions, showing that keyword search volume alone is not sufficient for prioritizing pages.
- Pages with different trend directions have similar median word counts, suggesting that content length alone does not explain declining performance.
- The starter pipeline achieved a much higher Precision@50 with the Random Forest model than with the hand-written baseline rule, indicating that combining multiple observable signals can improve page prioritization.

These observations suggest that using multiple signals together is likely to produce better recommendations than relying on a single metric.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("=" * 50)
print("Dataset Overview")
print("=" * 50)

print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"\nCorrelation between search_volume and impressions_90d: {corr:.3f}")

print("\nMedian Word Count by Trend Direction")
print(df.groupby("trend_direction")["word_count"].median().round())

print("\nAverage CTR by Position Tier")
print(df.groupby("position_tier")["ctr"].mean().round(4))
```
